```python
#Create gas space-heaing consumption
import numpy as np

df["gas_heating_kbtu"] = 0.0

for i in range(1, 6):
    source_col = f"Energy Source : Energy Source {i}"
    heating_col = f"Space Heating : Energy Source {i} (kBtu)"

    source_is_gas = df[source_col].eq("Natural Gas")
    heating_value = pd.to_numeric(
        df[heating_col],
        errors="coerce"
    ).fillna(0)

    df["gas_heating_kbtu"] += heating_value.where(
        source_is_gas,
        0
    )

df["gas_heating_mmbtu"] = df["gas_heating_kbtu"] / 1_000

#create boiler indicator
df["has_boiler"] = (
    df[heating_type_cols]
    .fillna("")
    .astype(str)
    .apply(
        lambda row: row.str.contains(
            "boiler",
            case=False,
            na=False
        ).any(),
        axis=1
    )
)

#create gas boiler-candidate list
gas_boiler_candidates = df[
    (df["gas_heating_kbtu"] > 0) &
    (df["has_boiler"])
].copy()

#convert submission date to datetime
gas_boiler_candidates["Submission Date"] = pd.to_datetime(
    gas_boiler_candidates["Submission Date"],
    errors="coerce",
    utc=True
)

# Sort newest reports first
gas_boiler_candidates = gas_boiler_candidates.sort_values(
    "Submission Date",
    ascending=False,
    na_position="last"
)

# Keep one report per property
gas_boiler_unique = (
    gas_boiler_candidates
    .drop_duplicates(subset="Property ID", keep="first")
    .copy()
)


#inspect resutlts
candidate_cols = [
    "Property ID",
    "Property Name",
    "Total Floor Area",
    "gas_heating_kbtu",
    "gas_heating_mmbtu",
    "has_boiler"
] + energy_source_cols + heating_type_cols

gas_boiler_candidates[candidate_cols].head(20)

#count candidates
print("Candidate report rows:", len(gas_boiler_candidates))
print(
    "Candidate properties:",
    gas_boiler_candidates["Property ID"].nunique()
)

#rank candidates by gas-heating amount
top_gas_boiler_candidates = (
    gas_boiler_candidates
    .sort_values("gas_heating_mmbtu", ascending=False)
)

print("Unique candidate properties:", gas_boiler_unique["Property ID"].nunique())

top_gas_boiler_candidates = (
    gas_boiler_unique
    .sort_values("gas_heating_mmbtu", ascending=False)
)

top_gas_boiler_candidates = (
    gas_boiler_unique
    .sort_values("gas_heating_mmbtu", ascending=False)
)

top_gas_boiler_candidates = (
    gas_boiler_unique
    .sort_values("gas_heating_mmbtu", ascending=False)
)

top_gas_boiler_candidates = (
    gas_boiler_unique
    .sort_values("gas_heating_mmbtu", ascending=False)
)

top_gas_boiler_candidates = (
    gas_boiler_unique
    .sort_values("gas_heating_mmbtu", ascending=False)
)

top_gas_boiler_candidates = (
    gas_boiler_unique
    .sort_values("gas_heating_mmbtu", ascending=False)
)

print("Before:", len(gas_boiler_candidates))
print("After:", len(gas_boiler_unique))
print(
    "Unique IDs:",
    gas_boiler_unique["Property ID"].nunique()
)


top_gas_boiler_candidates[candidate_cols].head(20)
```

```python
#find how many tons of Co2 would be saved if all gas boilers were replaced with heat pumps
#using 80% boiler efficiency and 300% heat pump efficiency (66.4g CO2 per kBtu for gas, 28.2g CO2 per kBtu for heat pump)

GAS_G_CO2_PER_KBTU = 66.4
HEAT_PUMP_G_CO2_PER_KBTU = 28.2

gas_boiler_unique["gas_emissions_g"] = (
    gas_boiler_unique["gas_heating_kbtu"] *
    GAS_G_CO2_PER_KBTU
)

gas_boiler_unique["heat_pump_emissions_g"] = (
    gas_boiler_unique["gas_heating_kbtu"] *
    HEAT_PUMP_G_CO2_PER_KBTU
)

gas_boiler_unique["emissions_saved_g"] = (
    gas_boiler_unique["gas_emissions_g"] -
    gas_boiler_unique["heat_pump_emissions_g"]
)

gas_boiler_unique["emissions_saved_metric_tons"] = (
    gas_boiler_unique["emissions_saved_g"] / 1_000_000
)

baseline_tons = (
    gas_boiler_unique["gas_emissions_g"].sum() / 1_000_000
)

heat_pump_tons = (
    gas_boiler_unique["heat_pump_emissions_g"].sum() / 1_000_000
)

saved_tons = (
    gas_boiler_unique["emissions_saved_g"].sum() / 1_000_000
)

#units of metric tons of CO2 saved per year if all gas boilers were replaced with heat pumps
print("Natural-gas emissions:", baseline_tons)
print("Heat-pump emissions:", heat_pump_tons)
print("Emissions saved:", saved_tons)
print("Percent reduction:", saved_tons / baseline_tons * 100)
```

```python
#see how principle use types
building_type_counts = (
    gas_boiler_unique["Principal Use Type"]
    .fillna("Unknown")
    .value_counts()
)
number_of_building_types = (
    gas_boiler_unique["Principal Use Type"]
    .nunique(dropna=True)
)

print("Number of different building types:", number_of_building_types)

building_types = (
    gas_boiler_unique["Principal Use Type"]
    .fillna("Unknown")
    .value_counts()
    .rename_axis("Building Type")
    .reset_index(name="Number of Buildings")
)

building_types
```

```python
#calculate emissions cap for most common building types
# Emissions factors: metric tons CO2 per square foot
current_factors = {
    "Multifamily Housing": 0.00675,
    "K-12 School": 0.00675,
    "Office": 0.00758
}

future_factors = {
    "Multifamily Housing": 0.003346640,
    "K-12 School": 0.002230588,
    "Office": 0.002690852
}

# Match each building to its factor
gas_boiler_unique["current_factor"] = (
    gas_boiler_unique["Principal Use Type"]
    .map(current_factors)
)

gas_boiler_unique["future_factor"] = (
    gas_boiler_unique["Principal Use Type"]
    .map(future_factors)
)

# Ensure floor area is numeric
floor_area = pd.to_numeric(
    gas_boiler_unique["Total Floor Area"],
    errors="coerce"
)

# Calculate current and future emissions caps
gas_boiler_unique["current_emissions_cap_tco2"] = (
    floor_area *
    gas_boiler_unique["current_factor"]
)

gas_boiler_unique["future_emissions_cap_tco2"] = (
    floor_area *
    gas_boiler_unique["future_factor"]
)

# Current natural-gas emissions
GAS_G_CO2_PER_KBTU = 66.4

gas_boiler_unique["current_emissions_tco2"] = (
    gas_boiler_unique["gas_heating_kbtu"] *
    GAS_G_CO2_PER_KBTU /
    1_000_000
)

# Difference from each cap
gas_boiler_unique["difference_from_current_cap_tco2"] = (
    gas_boiler_unique["current_emissions_tco2"] -
    gas_boiler_unique["current_emissions_cap_tco2"]
)

gas_boiler_unique["difference_from_future_cap_tco2"] = (
    gas_boiler_unique["current_emissions_tco2"] -
    gas_boiler_unique["future_emissions_cap_tco2"]
)

# Above/below status
gas_boiler_unique["current_cap_status"] = (
    gas_boiler_unique["difference_from_current_cap_tco2"]
    .apply(
        lambda value:
        "Above cap" if value > 0
        else "Below cap" if value < 0
        else "At cap"
    )
)

gas_boiler_unique["future_cap_status"] = (
    gas_boiler_unique["difference_from_future_cap_tco2"]
    .apply(
        lambda value:
        "Above cap" if value > 0
        else "Below cap" if value < 0
        else "At cap"
    )
)

gas_boiler_unique["difference_from_current_cap_tco2"] = (
    gas_boiler_unique["current_emissions_tco2"] -
    gas_boiler_unique["current_emissions_cap_tco2"]
)

gas_boiler_unique["difference_from_future_cap_tco2"] = (
    gas_boiler_unique["current_emissions_tco2"] -
    gas_boiler_unique["future_emissions_cap_tco2"]
)

```

```python
cap_table = gas_boiler_unique[
    [
        "Property ID",
        "Property Name",
        "Principal Use Type",
        "Total Floor Area",
        "current_emissions_tco2",
        "current_emissions_cap_tco2",
        "current_cap_status",
        "future_emissions_cap_tco2",
        "future_cap_status",
        "difference_from_future_cap_tco2"
    ]
].sort_values(
    "difference_from_future_cap_tco2",
    ascending=False,
    na_position="last"
)

cap_table.head(20)
```

```python
cap_table["future_cap_status"].value_counts(dropna=False)

num_above_future_cap = (
    cap_table["future_cap_status"] == "Above cap"
).sum()

print("Buildings above future cap:", num_above_future_cap)

num_below_future_cap = (
    cap_table["future_cap_status"] == "Below cap"
).sum()

print("Buildings below future cap:", num_below_future_cap)

cap_table[
    cap_table["current_emissions_cap_tco2"].notna()
]["current_cap_status"].value_counts()
```

```python
HEAT_PUMP_G_CO2_PER_KBTU = 28.2

gas_boiler_unique["heat_pump_emissions_tco2"] = (
    gas_boiler_unique["gas_heating_kbtu"] *
    HEAT_PUMP_G_CO2_PER_KBTU /
    1_000_000
)

gas_boiler_unique["emissions_saved_by_heat_pump_tco2"] = (
    gas_boiler_unique["current_emissions_tco2"] -
    gas_boiler_unique["heat_pump_emissions_tco2"]
)

emissions_ranking = cap_table.sort_values(
    "emissions_saved_by_heat_pump_tco2",
    ascending=False,
    na_position="last"
)
```

```python
emissions_ranking[
    [
        "Property ID",
        "Property Name",
        "Principal Use Type",
        "Total Floor Area",
        "current_emissions_tco2",
        "heat_pump_emissions_tco2",
        "emissions_saved_by_heat_pump_tco2",
    ]
].head(20)
```

```python
building_type_summary = (
    gas_boiler_unique
    .groupby("Principal Use Type", dropna=False)
    .agg(
        buildings=("Property ID", "nunique"),
        total_floor_area=("Total Floor Area", "sum"),
        average_floor_area=("Total Floor Area", "mean"),
        total_emissions_saved=(
            "emissions_saved_by_heat_pump_tco2",
            "sum"
        ),
        average_emissions_saved=(
            "emissions_saved_by_heat_pump_tco2",
            "mean"
        ),
        median_emissions_saved=(
            "emissions_saved_by_heat_pump_tco2",
            "median"
        )
    )
    .sort_values(
        "total_emissions_saved",
        ascending=False
    )
)

building_type_summary
```

```python
gas_boiler_unique["floor_area_group"] = pd.cut(
    gas_boiler_unique["Total Floor Area"],
    bins=[0, 50_000, 100_000, 250_000, 500_000, float("inf")],
    labels=[
        "25,000–50,000",
        "50,001–100,000",
        "100,001–250,000",
        "250,001–500,000",
        "500,000+"
    ]
)

floor_area_summary = (
    gas_boiler_unique
    .groupby("floor_area_group", observed=False)
    .agg(
        buildings=("Property ID", "nunique"),
        total_emissions_saved=(
            "emissions_saved_by_heat_pump_tco2",
            "sum"
        ),
        average_emissions_saved=(
            "emissions_saved_by_heat_pump_tco2",
            "mean"
        ),
        median_emissions_saved=(
            "emissions_saved_by_heat_pump_tco2",
            "median"
        )
    )
)

floor_area_summary
```
